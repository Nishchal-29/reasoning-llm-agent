```python
import sys
import os
import shutil

print(f"sys.path BEFORE modification:\n{os.linesep.join(sys.path)}")
current_cwd = os.getcwd()
removed_cwd_from_sys_path = False
if current_cwd in sys.path:
    sys.path.remove(current_cwd)
    removed_cwd_from_sys_path = True
    print(f"Temporarily removed current working directory from sys.path: {current_cwd}")

local_datasets_path = os.path.join(current_cwd, "datasets")
print(f"Checking for conflicting local 'datasets' directory at: {local_datasets_path}")
if os.path.exists(local_datasets_path) and os.path.isdir(local_datasets_path):
    print("Conflicting 'datasets' directory found. Attempting to rename it to './_datasets_backup' to avoid import conflicts.")
    try:
        shutil.move(local_datasets_path, os.path.join(current_cwd, "_datasets_backup"))
        print(f"Successfully renamed '{local_datasets_path}' to '{os.path.join(current_cwd, '_datasets_backup')}'")
    except Exception as e:
        print(f"Could not rename '{local_datasets_path}': {e}. Proceeding without renaming, but expect potential issues.")
else:
    print(f"No conflicting local 'datasets' directory found at {local_datasets_path}.")

if 'datasets' in sys.modules:
    print("Removing 'datasets' from sys.modules to force a fresh import.")
    del sys.modules['datasets']
for module_name in list(sys.modules.keys()):
    if module_name.startswith('datasets.'):
        print(f"Removing submodule '{module_name}' from sys.modules.")
        del sys.modules[module_name]

print(f"sys.path AFTER modification (before unsloth import):\n{os.linesep.join(sys.path)}")
from unsloth import FastLanguageModel
if removed_cwd_from_sys_path:
    sys.path.insert(0, current_cwd)
    print(f"Restored current working directory to sys.path: {current_cwd}")

print("Loading model and adapter")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="./outputs/sft_reasoning",
    max_seq_length=1024,
    load_in_4bit=True,
)

FastLanguageModel.for_inference(model)
system_message = (
    "You are a precise mathematical reasoning agent. "
    "Show your step-by-step reasoning in <think> tags, "
    "then provide your final answer in <final_answer> tags."
)

test_question = "A store sells apples for $2 each and oranges for $3 each. If Sarah buys 4 apples and 3 oranges, and pays with a $50 bill, how much change does she get?"
prompt = (
    f"<|im_start|>system\n{system_message}\n<|im_end|>\n"
    f"<|im_start|>user\n{test_question}\n<|im_end|>\n"
    f"<|im_start|>assistant\n"
)

inputs = tokenizer([prompt], return_tensors="pt").to("cuda")
print("\nGenerating response...\n")
print("-" * 50)
outputs = model.generate(
    **inputs,
    max_new_tokens=256,
    use_cache=True,
    temperature=0.3,
)

input_length = inputs.input_ids.shape[1]
generated_tokens = outputs[0][input_length:]
response = tokenizer.decode(generated_tokens, skip_special_tokens=True)
print(response)
print("-" * 50)
```
